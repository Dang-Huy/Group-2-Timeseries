```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('g2data_final.csv')
df
```

```python
df = df.set_index('date')
df.index = pd.to_datetime(df.index)
df_monthly = df['dlog_GoldPrice']
df_monthly = df_monthly.resample('ME').mean()
df_monthly
```

```python
import plotly.express as px
fig = px.line(df_monthly, x=df_monthly.index, y='dlog_GoldPrice', title='Monthly Average of Gold Price')
fig.show()
```
